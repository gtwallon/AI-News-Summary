```python
# import libraries
import newspaper
from newspaper import Article
import requests
```

```python
# Use a free API to retrieve news articles that match a provided query
query = "(sustainability AND business AND climate AND impact)"  #the article titles, descriptions, and content are all searched for their fit with this query
start_date = "2025-04-23"
end_date = "2025-04-30"
sort_by = "relevancy"  #choose either relevancy, popularity, publishedAt
NEWS_API_KEY= # <your free NewsAPI key>

url = f"https://newsapi.org/v2/everything?q={query}&from={start_date}&to={end_date}&sortBy={sort_by}&apiKey={NEWS_API_KEY}"
r = requests.get(url)
```

```python
# look a data
articles = r.json()['articles']
articles[:3]
```

```python
# total number of articles retrieved
len(articles)
```

```python
# cleaning the 'source name' field
for article in articles:
    article['source'] = article['source']['name']
```

```python
# look at cleaned data
articles[0]
```

```python
import pandas as pd
import numpy as np

# create a dataframe - each row represents an article
df = pd.DataFrame.from_dict(articles)
```

```python
# subset to only the first 10 rows for testing
head = df.head(10)
head.head()
```

```python
# filter data further
keywords = ['sustainability', 'sustainable', 'business', 'climate', 'impact']

filtered_df = head[
    (head['title'].str.contains('|'.join(keywords), case=False)) |
    (head['description'].str.contains('|'.join(keywords), case=False)) |
    (head['content'].str.contains('|'.join(keywords), case=False))
    ]
```

```python
import os
import dotenv
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(), override=True)

os.environ.get('OPENAI_API_KEY')

from openai import OpenAI

token = os.environ.get('GITHUB_TOKEN')
endpoint = "https://models.inference.ai.azure.com"
model_name = "gpt-4o-mini"

client = OpenAI(
    base_url=endpoint,
    api_key=token
)
```

```python
from pydantic import BaseModel

class ArticleEval(BaseModel):
    rating: int
    rationale: str
```

```python
# rate the articles based on their relevance using AI
responses = []
topic = "innovative business solutions to climate change."

for i in range(filtered_df.shape[0]):

  title = filtered_df.iloc[i]['title']
  descr = filtered_df.iloc[i]['description']
  base_messages = [
      {"role": "developer", "content": "You are a helpful AI assistant who will be asked to rate articles on a scale from 1 to 10 based on "
                                        "their relevance to a provided topic of interest. You will be provided the article title and description, "
                                        "as well as the topic of interest"},
      {"role": "user", "content": (f"Please provide a rating for the article with the following title and description based on how relevant "
                                  f"it is to {topic}: \n\n Article Title: {title} \n\n Article Description:{descr}")},
      {"role": "user", "content": "Please also provide two sentances explaining how you decided on this rating."}
  ]

  reply = client.beta.chat.completions.parse(
                messages=base_messages,
                temperature=1.0,
                top_p=1.0,
                max_tokens=1000,
                model=model_name,
                response_format=ArticleEval
            )
  responses.append(reply.choices[0].message.content)
```

```python
# format responses
import ast

new_list = []
for r in responses:
  new_list.append(ast.literal_eval(r))

#create a new dataframe from responses
responses_df = pd.DataFrame.from_dict(new_list)

# concatenate the two dataframes
next_df = pd.concat([filtered_df.reset_index(), responses_df], axis=1)
next_df.drop('index', axis=1, inplace=True)
next_df
```

```python
# sort dataframe by rating
next_df = next_df.sort_values(by='rating', ascending=False)
```

```python
# summarize whole articles
from newspaper import Config

summaries = []
for i in range(5):
  article_url = next_df.iloc[i]['url']
  text = Article(article_url)
  text.download()
  try:
    text.parse()
  except:
    user_agent = 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_11_5) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/50.0.2661.102 Safari/537.36'
    config = Config()
    config.browser_user_agent = user_agent
    text = Article(article_url, config=config)
    text.download()
    text.parse()

  ai_input = text.text
  output_num_sents = 5
  base_messages = [
        {"role": "developer", "content": "You are a helpful AI assistant who will be asked to summarize articles."},
        {"role": "user", "content": f"Please summarize the following article using {output_num_sents} sentances:\n\n {ai_input}"}
    ]

  reply = client.chat.completions.create(
                messages=base_messages,
                temperature=1.0,
                top_p=1.0,
                max_tokens=1000,
                model=model_name
            )
  summaries.append({'article_idx':i, 'summary':reply.choices[0].message.content})
```
